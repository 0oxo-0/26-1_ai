```python
# pandas는 표 형태 데이터 처리, numpy는 수치 계산에 사용합니다.
import pandas as pd
import numpy as np

# 데이터 경로
# 사용할 CSV 파일의 경로를 지정합니다.
path = '/content/drive/MyDrive/notebook/diabetes.csv'

# CSV 파일 읽기
# CSV 파일을 읽어 pandas DataFrame으로 만듭니다.
df = pd.read_csv(path)

# Outcome 열 제거
# 회귀 실습에서는 분류용 정답 컬럼인 Outcome을 제외합니다.
df2 = df.drop('Outcome', axis=1)

# 예측할 값 (BMI)
# 예측해야 할 정답(label 또는 target) 컬럼을 y에 저장합니다.
y = df2['BMI']

# 입력 데이터 (BMI 제외)
# 모델 입력에 사용할 특징(feature) 데이터를 X에 저장합니다.
X = df2.drop('BMI', axis=1)

# 데이터 분리
# 데이터를 학습용과 테스트용으로 나누기 위한 함수를 불러옵니다.
from sklearn.model_selection import train_test_split

# 전체 데이터 중 75%는 학습, 25%는 테스트에 사용합니다.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0
)

from sklearn.metrics import mean_squared_error

# 선형 회귀 모델
# 선형 회귀는 연속적인 숫자 값을 예측하는 기본 회귀 모델입니다.
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 선형 회귀 모델 객체를 생성합니다.
model = LinearRegression()

# 모델 학습
# 학습 데이터로 모델을 훈련합니다.
model.fit(X_train, y_train)

# 예측
# 테스트 데이터에 대한 예측값을 생성합니다.
ly_preds = model.predict(X_test)

# sklearn으로 MSE 계산
# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print('평균제곱오차', mean_squared_error(ly_preds, y_test))


# numpy로 MSE 계산
# numpy 연산으로 MSE를 직접 계산하는 함수입니다.
def mse_np(actual, predicted):
    return np.mean((np.array(actual) - np.array(predicted)) ** 2)

print('평균제곱오차', mse_np(ly_preds, y_test))


# 직접 MSE 계산
# 반복문과 sum을 이용해 MSE 계산 과정을 직접 구현합니다.
def mse(actual, predicted):
    sum_square_error = sum((a - p) ** 2 for a, p in zip(actual, predicted))
    mean_square_error = sum_square_error / len(actual)
    return mean_square_error

print('평균제곱오차', mse(ly_preds, y_test))


# 그래프 출력
# matplotlib은 그래프를 그릴 때 사용합니다.
import matplotlib.pyplot as plt

# 그래프가 그려질 전체 화면 크기를 설정합니다.
plt.figure(figsize=(10,5))

# 실제값 vs 예측값
# 실제값과 예측값이 얼마나 비슷한지 산점도로 확인합니다.
plt.scatter(y_test, ly_preds)

plt.xlabel("Actual BMI")
plt.ylabel("Predicted BMI")

plt.title("BMI Prediction Result")

plt.show()
```

```python
# 선형 회귀는 연속적인 숫자 값을 예측하는 기본 회귀 모델입니다.
from sklearn.linear_model import LinearRegression

# 선형 회귀 모델 객체를 생성합니다.
lr = LinearRegression()

# 학습 데이터로 모델을 훈련합니다.
lr.fit(X_train, y_train)

# 테스트 데이터에 대한 예측값을 생성합니다.
pred_lr = lr.predict(X_test)

print("\n--- Linear Regression ---")
# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print("MSE:", mean_squared_error(y_test, pred_lr))
```

```python
# 결정 트리 회귀는 데이터를 조건별로 나누어 연속값을 예측합니다.
from sklearn.tree import DecisionTreeRegressor

# 결정 트리 회귀 모델을 생성합니다.
dt = DecisionTreeRegressor(random_state=0)

# 학습 데이터로 모델을 훈련합니다.
dt.fit(X_train, y_train)

# 테스트 데이터에 대한 예측값을 생성합니다.
pred_dt = dt.predict(X_test)

print("\n--- Decision Tree Regressor ---")
# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print("MSE:", mean_squared_error(y_test, pred_dt))
```
