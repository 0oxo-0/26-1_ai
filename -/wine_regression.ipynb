```python

# pandas는 표 형태 데이터 처리, numpy는 수치 계산에 사용합니다.
import pandas as pd
import numpy as np

# 데이터 불러오기
# 사용할 CSV 파일의 경로를 지정합니다.
path = '/content/drive/MyDrive/Colab Notebooks/wine.csv'
# CSV 파일을 읽어 pandas DataFrame으로 만듭니다.
df = pd.read_csv(path, index_col=0)

# df = df.drop('Unnamed: 0', axis=1)

# 예측해야 할 정답(label 또는 target) 컬럼을 y에 저장합니다.
y = df['Alcohol']

# 모델 입력에 사용할 특징(feature) 데이터를 X에 저장합니다.
X = df.drop('Alcohol', axis=1)


# 데이터를 학습용과 테스트용으로 나누기 위한 함수를 불러옵니다.
from sklearn.model_selection import train_test_split

# 전체 데이터 중 75%는 학습, 25%는 테스트에 사용합니다.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0
)
```

```python
# 선형 회귀는 연속적인 숫자 값을 예측하는 기본 회귀 모델입니다.
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 선형 회귀 모델 객체를 생성합니다.
model = LinearRegression()

# 학습 데이터로 모델을 훈련합니다.
model.fit(X_train, y_train)

# 테스트 데이터에 대한 예측값을 생성합니다.
pred = model.predict(X_test)

# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print("MSE:", mean_squared_error(y_test, pred))
```

```python
# 결정 트리 회귀는 데이터를 조건별로 나누어 연속값을 예측합니다.
from sklearn.tree import DecisionTreeRegressor

# 결정 트리 회귀 모델을 생성합니다.
dt = DecisionTreeRegressor(random_state=0)

# 학습 데이터로 모델을 훈련합니다.
dt.fit(X_train, y_train)

# 테스트 데이터에 대한 예측값을 생성합니다.
pred_dt = dt.predict(X_test)

print("\n--- Decision Tree Regressor ---")
# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print("MSE:", mean_squared_error(y_test, pred_dt))
```

```python
# 랜덤 포레스트 회귀는 여러 트리의 예측을 평균내어 안정적인 예측을 만듭니다.
from sklearn.ensemble import RandomForestRegressor

# 랜덤 포레스트 회귀 모델을 생성합니다.
rf = RandomForestRegressor(random_state=0)

# 학습 데이터로 모델을 훈련합니다.
rf.fit(X_train, y_train)

# 테스트 데이터에 대한 예측값을 생성합니다.
pred_rf = rf.predict(X_test)

print("\n--- Random Forest Regressor ---")
# MSE는 실제값과 예측값의 차이를 제곱해 평균낸 회귀 오차입니다.
print("MSE:", mean_squared_error(y_test, pred_rf))
```
